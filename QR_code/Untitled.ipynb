```python
# Create qr code 
import qrcode
url = input("Enter URL")
if url.startswith("http://") or url.startswith("https://"):
                                               qr = qrcode.QRCode(version=1, error_correction=qrcode.constants.ERROR_CORRECT_H,
                                               box_size=10,
                                               border=4)
                                               qr.add_data(url)
                                               qr.make()

                                               image=qr.make_image(fill_color="blue", back_color="red")
                                               image.save("ThirdQr.png")
else:
    print("Invalid url")
                                               
                                               
```

```python
# Create a UPI QR Code
import qrcode

upi_id = input("Enter UPI ID:")
name = input("Enter Name:")

upi_url = f"upi://pay?pa={upi_id}&pn={name}"

qr = qrcode.make(upi_url)

qr.save("surya.png")
```

```python
# Create qr code for Upi with specific amount 
import qrcode

upi_id = input("Enter UPI ID:")
name = input("Enter Name:")
amount = int(input("Enter Amount"))
upi_url = f"upi://pay?pa={upi_id}&pn={name}&am-{amount}"

qr = qrcode.make(upi_url)

qr.save("surya.png")
```
